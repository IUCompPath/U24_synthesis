```python
import os
import nibabel as nib
import random

root = "/mnt/d/Datasets/inbreast/inbreast_clean/train/original"

def sample_shapes(class_name, n=5):
    class_dir = os.path.join(root, class_name)
    cases = os.listdir(class_dir)
    random.shuffle(cases)

    print(f"\n=== {class_name.upper()} ===")
    
    for case in cases[:n]:
        case_dir = os.path.join(class_dir, case)
        nii_files = [f for f in os.listdir(case_dir) if f.endswith(".nii.gz")]
        if not nii_files:
            continue

        path = os.path.join(case_dir, nii_files[0])
        img = nib.load(path).get_fdata()

        print(f"{case}: shape={img.shape}, ndim={img.ndim}")

# run it
sample_shapes("benign", 5)
sample_shapes("malignant", 5)
```
